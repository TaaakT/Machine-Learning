```python
df['total_spend']=df['TV']+df['newspaper']+df['radio']
```

```python
X= df['total_spend']
y= df['sales']
```

```python
w,b=np.polyfit(X,y,1)
```

```python
spend_for_prediction=np.linspace(0,500,100)
```

```python
pred_sales = w*spend_for_prediction+b
pred_sales,spend_for_prediction
```

```python
plt.figure(figsize=(10,5))
plt.scatter(df['total_spend'],df['sales'])
plt.plot(spend_for_prediction,pred_sales,'r-+')
plt.axis([0,500,0,30])
```

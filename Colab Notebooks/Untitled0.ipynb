```python
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.2)
```

```python
cat_pipeline = Pipeline([
    ('impute',SimpleImputer(strategy='most_frequent')),
    ('onehot',OneHotEncoder(sparse_output=False))
])
num_pipeline = Pipeline([
    ('impute',SimpleImputer(strategy='median'))
])
```

```python
num_features=X_train.select_dtypes(include=[np.number]).columns
cat_features=X_train.select_dtypes(exclude=[np.number]).columns
```

```python
cat_features
```

```python
transformer = ColumnTransformer([
    ('cat',cat_pipeline,cat_features),
    ('num',num_pipeline,num_features)
], remainder='passthrough')
```

```python
X_train_transformed = transformer.fit_transform(X_train)
X_train_transformed
```

```python
columns=transformer.get_feature_names_out()
columns
```

```python
X_train_transformed_df = pd.DataFrame(data=X_train_transformed,columns=columns)
X_train_transformed_df
```

```python
  X_test_transformed=transformer.transform(X_test)
```

```python
X_test_transformed_df = pd.DataFrame(data=X_test_transformed,columns=columns)
X_test_transformed_df
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
```

```python
lr= LinearRegression()
rfc = RandomForestRegressor()
```

```python
lr.fit(X_train_transformed_df,y_train)
```

```python
print(f'train score: {lr.score(X_train_transformed_df,y_train)}')
print(f'test score: {lr.score(X_test_transformed_df,y_test)}')
```

```python
rfc.fit(X_train_transformed_df,y_train)
```

```python
print(f'train score: {rfc.score(X_train_transformed_df,y_train)}')
print(f'test score: {rfc.score(X_test_transformed_df,y_test)}')
```

```python
df.dropna(subset=['price_usd'], inplace=True)
```

```python
X=df.drop('price_usd',axis=1)
y=df['price_usd'].copy()
```

```python
from sklearn.model_selection import train_test_split
import numpy as np
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2)
```

```python
df.isna().sum()
```

```python
df.dropna(subset=['price_usd'], inplace=True)
```

```python
num_features=X_train.select_dtypes(include=[np.number]).columns
cat_features=X_train.select_dtypes(exclude=[np.number]).columns
```

```python
cat_features
```

```python
cat_pipeline = Pipeline([
    ('impute',SimpleImputer(strategy='most_frequent')),
    ('onehot',OneHotEncoder(handle_unknown='ignore'))
])
num_pipeline = Pipeline([
    ('impute',SimpleImputer(strategy='median')),
    ('scale',StandardScaler())
])
transformer = ColumnTransformer([
    ('cat',cat_pipeline,cat_features),
    ('num',num_pipeline,num_features)
], remainder='passthrough')

full_pipeline = Pipeline([
    ('preprocessing',transformer),
    ('estimator',LinearRegression())
])
```

```python
full_pipeline.fit(X_train,y_train)
```

```python
y_test.isna().sum()
```

```python
print(f'train score: {full_pipeline.score(X_train,y_train)}')
print(f'test score: {full_pipeline.score(X_test,y_test)}')
```

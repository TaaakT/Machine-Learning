```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('/content/drive/MyDrive/DATA/mouse_viral_study.csv')
```

```python
from svm_margin_plot import plot_svm_boundary
```

```python
df.head()
```

```python
X=df.drop('Virus Present',axis=1)
y=df['Virus Present']
```

```python
from sklearn.svm import SVC
```

```python
svm=SVC(kernel='linear',C=100)
```

```python
svm.fit(X,y)
```

```python
plot_svm_boundary(svm,X,y)
```

```python
svm=SVC(kernel='linear',C=0.1)
svm.fit(X,y)
plot_svm_boundary(svm,X,y)
```

```python
svm=SVC(kernel='linear',C=0.01)
svm.fit(X,y)
plot_svm_boundary(svm,X,y)
```

```python
svm=SVC(kernel='poly',C=0.1,degree=3)
svm.fit(X,y)
plot_svm_boundary(svm,X,y)
```

```python
svm=SVC(kernel='rbf',C=0.1)
svm.fit(X,y)
plot_svm_boundary(svm,X,y)
```

```python
svm=SVC(kernel='rbf',C=0.1,gamma='auto')#a\gamma=auto = 1/n_features=1/2
svm.fit(X,y)
plot_svm_boundary(svm,X,y)
```

```python
svm=SVC(kernel='rbf',C=0.1,gamma=0.5)
svm.fit(X,y)
plot_svm_boundary(svm,X,y)
```

```python
svm=SVC(kernel='rbf',C=0.1,gamma=1)#gamma>  ->   more support vectors and more chance for overfitting
svm.fit(X,y)
plot_svm_boundary(svm,X,y)
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
model=SVC()
```

```python
params={'kernel':['linear','rbf'],
         'C':[0.01,0.1,1]}
```

```python
grid=GridSearchCV(model,param_grid=params)
```

```python
grid.fit(X,y)
```

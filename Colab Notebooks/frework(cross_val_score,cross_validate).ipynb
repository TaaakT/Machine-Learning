```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('/content/drive/MyDrive/Advertising.csv') #datani goturduk
```

```python
#datani temizlemek lazim deyil(evvelceden temizlenmis datadi)
#X ve y boluruk
X=df.drop('sales',axis=1)
y=df['sales']
```

```python
from sklearn.model_selection import train_test_split
#cross validation ucun train teste boluruk (cv train datasinda olacaq, test yalniz sonda modeli qiymetlendirmek ucun istifade olunacaq)
X_train, X_test,y_train,y_test = train_test_split(X,y,test_size=0.3,random_state=101)
```

```python
from sklearn.preprocessing import StandardScaler
#scaling edirik (linear modeller ucun arzuolunandir sum((x-u)/s) u her sutunun ortalamasidi, s standart yayimlamadi )
scaler=StandardScaler()
scaler.fit(X_train) #fit'e yalniz train datani veririk
X_train=scaler.transform(X_train)
```

```python
X_test=scaler.transform(X_test)
```

```python
from sklearn.linear_model import Ridge
#modelimize bilerek boyuk alpha veririk
model=Ridge(alpha=100)
```

```python
from sklearn.model_selection import cross_val_score
#cross validation 5 hisseden ibaret olacaq ve her defe score taapilacaq
scores= cross_val_score(model,X_train,y_train,scoring='neg_mean_squared_error',cv=5)
```

```python
scores
```

```python
abs(scores.mean())
```

```python
model=Ridge(alpha=1)
scores= cross_val_score(model,X_train,y_train,scoring='neg_mean_squared_error',cv=5)
abs(scores.mean())
```

Cross validation'dan sonra model fit olunmur, biz yalniz modelin 5 train test scorlarina baxiriq. scorlarina baxaraq biz overfitting olub ya ol,ayiq melumatini aliriq

```python
# modeli fit edek
model.fit(X_train,y_train)
```

```python
y_pred=model.predict(X_test)
```

```python
from sklearn.metrics import mean_squared_error
mse=mean_squared_error(y_test,y_pred)
mse
#final scorumuzu modelin gormeyen test datasi ile edirik
```

# Cross_validate()

```python
X=df.drop('sales',axis=1)
y=df['sales']

from sklearn.model_selection import train_test_split
X_train, X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=101)

from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
scaler.fit(X_train)
X_train=scaler.transform(X_train)
X_test=scaler.transform(X_test)
```

```python
from sklearn.model_selection import cross_validate
```

```python
model= Ridge(alpha=100)
```

```python
scores=cross_validate(model,X_train,y_train,scoring=['neg_mean_squared_error','neg_mean_absolute_error'],cv=10)
```

```python
pd.DataFrame(scores)
```

```python
pd.DataFrame(scores).mean()
```

```python
model=Ridge(alpha=1)
scores=cross_validate(model,X_train,y_train,scoring=['neg_mean_squared_error','neg_mean_absolute_error'],cv=10)
```

```python
pd.DataFrame(scores)
```

```python
pd.DataFrame(scores).mean()
```

```python
model.fit(X_train,y_train)
y_pred=model.predict(X_test)
```

```python
from sklearn.metrics import mean_absolute_error
```

```python
mse=mean_squared_error(y_test,y_pred)
mae=mean_absolute_error(y_test,y_pred)
```

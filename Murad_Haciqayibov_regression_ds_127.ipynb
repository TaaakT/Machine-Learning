```python
from sklearn.preprocessing import LabelEncoder
encoder=LabelEncoder()
df['Yanacaq']=encoder.fit_transform(df['Yanacaq'])
df['Ötürücü']=encoder.fit_transform(df['Ötürücü'])
df['Sürətlər qutusu']=encoder.fit_transform(df['Sürətlər qutusu'])
df['Vəziyyəti']=encoder.fit_transform(df['Vəziyyəti'])
df['Yerlərin sayı']=encoder.fit_transform(df['Yerlərin sayı'])
```

```python
df['Yanacaq']=df['Yanacaq'].astype('int32')
df['Ötürücü']=df['Ötürücü'].astype('int32')
df['Sürətlər qutusu']=df['Sürətlər qutusu'].astype('int32')
df['Vəziyyəti']=df['Vəziyyəti'].astype('int32')
df['Yerlərin sayı']=df['Yerlərin sayı'].astype('int32')
```

```python
df=pd.get_dummies(df,drop_first=True)
```

```python
X=df.drop(['price_x'],axis=1)
y=df['price_x']
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
from sklearn.linear_model import LinearRegression

model_1=LinearRegression()
model_1.fit(X_train,y_train)
```

```python
model_1.score(X_train,y_train)
```

```python
model_1.score(X_test,y_test)
```

```python
from sklearn.metrics import mean_squared_error
from sklearn.metrics import mean_absolute_error
y_pred=model_1.predict(X_test)
RMSE=np.sqrt(mean_squared_error(y_test,y_pred))
MAE = mean_absolute_error(y_test, y_pred)
print(RMSE)
print(MAE)
```
